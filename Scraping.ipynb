```python
from playwright.async_api import async_playwright

async with async_playwright() as p:
    browser = await p.chromium.launch(headless=True)
    page = await browser.new_page()

    await page.goto("https://www.charusat.ac.in/convocations")

    try:
        # Close announcement popup
        dialog = page.locator('dialog[aria-label="Announcement popup"]')
        if await dialog.is_visible():
            await dialog.locator("button").first.click()
    except Exception as e:
        print(f"Error occurred: {e}")

    # Click About Us
    # await page.locator("#KnowCHARUSAT-link").click()

    print(await page.locator("body").inner_text())

    await browser.close()
```
